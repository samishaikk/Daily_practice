Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Create an object of the class and perform some operations.

```python
class BankAccount():
    def __init__(self,account_number,balance):
        self.__accountnumber = account_number
        self.__balance = balance
    
    def deposit(self,amount):
        self.__balance += amount
        print(f'deposited: {amount}')
    
    def withdraw(self,amount):
        
        if amount<=self.__balance:
            self.__balance -= amount
            print(f'withdrawn amount: {amount}')
        
        else:
            return 'insufficient balance'
        
    def checkbalance(self):
        print(f'current balance: {self.__balance}')

account1 = BankAccount('acc11',1000)
account1.deposit(1000)
account1.checkbalance()


account1.deposit(5000)
account1.checkbalance()

account1.withdraw(3000)
account1.checkbalance()
```

Create a class named `Rectangle` with private attributes `length` and `width`. Use properties to get and set these attributes. Create an object of the class and test the properties.

```python
class Rectangle:
    
    def __init__(self,length,width):
        self.__length = length
        self.__width = width
        
        
    @property
    def length(self):
        return self.__length
    
    # setter for length
    
    @length.setter
    def length(self,value):
        self.__length = value
    
    @property
    def width(self):
        return self.__width
    
    # setter for width
    
    @width.setter
    def width(self,value):
        self.__width = value
        
r = Rectangle(10,5)

print(f'length: {r.length}')
print(f'width: {r.width}')

r.length = 18
r.width = 10

print(f'updated length: {r.length}')
print(f'updated width: {r.width}')
```

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    
    @abstractmethod
    def area(self):
        pass
    
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        
    def area (self):
        return 3.14 * self.radius * self.radius

class Square(Shape):
    def __init__(self,side):
        self.side = side
        
    def area(self):
        return self.side * self.side
    
    
c = Circle(9)
s = Square(7)

print(f'area of circle: {c.area()}')
print(f'area of square: {s.area()}')
```

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
        
    def __add__(self, other):
        return Vector(
            self.x+other.x,
            self.y+other.y
        )
    
    def __str__(self):
        return f'({self.x},{self.y})'
    
v1 = Vector(2,3)
v2 = Vector(4,5)

v3 = v1+v2

print('addition:',v3)
```

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python
class InsufficientBalanceError(Exception):
    pass

class BankAccount:
    
    def __init__(self,balance):
        self.balance = balance
    
    def withdraw(self,amount):
        
        if amount>self.balance:
            raise InsufficientBalanceError(
                'insufficient balance'
            )
        
        self.balance -= amount
        print('withdrawl done:',amount)
        print('remaining balance:', self.balance)
        
account = BankAccount(80000)

try:
    account.withdraw(9000)
except InsufficientBalanceError as e:
    print('error:',e)

try:
    account.withdraw(90000)
except InsufficientBalanceError as e:
    print('error:',e)
```

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
    def __init__(self,filename,mode):
        self.filename = filename
        self.mode = mode
        
    def __enter__(self):
        self.file = open(self.filename,self.mode)
        return self.file 
    
    def __exit__(self, exc_type, exc, tb):
        self.file.close()
        print('file closed')

with FileManager('sample.txt','r') as file:
    content = file.read()
    print(content)
```

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
    
    def __init__(self,value=0):
        self.value = value
    
    def add(self,num):
        self.value += num
        return self
    
    def subtract(self,num):
        self.value -= num
        return self
    
    def multiply(self,num):
        self.value *= num
        return self
    
    def divide(self,num):
        self.value /= num
        return self
    
    def get_result(self):
        return self.value
calc = Calculator()

result = calc.add(10).multiply(2).subtract(5).divide(3).get_result()

print('output:', result)
```
